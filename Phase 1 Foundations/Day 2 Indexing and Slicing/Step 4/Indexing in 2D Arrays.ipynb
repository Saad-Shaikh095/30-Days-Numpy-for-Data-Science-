```python
import numpy as np
```

```python
b = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
```

```python
b
```

# Access Single Element

```python
b[0, 1]   # 20
b[2, 2]   # 90
```

# Access Whole Row

```python
b[1]      # [40 50 60]
```

# Access whole Column


```python
b[:, 1]   # [20 50 80]
```

# Important rule

*   First → rows

*   Second → columns
