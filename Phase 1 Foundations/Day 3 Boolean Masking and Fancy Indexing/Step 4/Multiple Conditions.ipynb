```python
import numpy as np
```

```python
marks = np.array([45, 60, 72, 88, 35, 90])

passed = marks[marks >= 50]
print(passed)
```

# AND Condition

```python
marks[(marks >= 50) & (marks < 80)]
```

# OR Condition

```python
marks[(marks < 40) | (marks > 85)]
```

# ⚠️ IMPORTANT:

* Use & and |

* NOT and, or

# 🧠 Memory trick:

> NumPy thinks in bitwise, not English.
