```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40])
view_arr = arr[1:3]
```

```python
view_arr[0] = 999
print(arr)
```

## 😱 Original array changes!

# 🧠 Meaning:

> A view shares memory.

```python
copy_arr = arr[1:3].copy()
copy_arr[0] = 111
```

## Original remains unchanged 😌

# 📌 Golden Rule

> If you don’t want surprises → use .copy()
